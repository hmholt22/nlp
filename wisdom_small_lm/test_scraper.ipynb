# Scraper Test — Song of Songs
Testing the generalized scraper on the smallest book (8 chapters) before running the full corpus.

```python
import sys
sys.path.insert(0, '.')
from scraper import scrape_book, BOOKS
```

```python
# Scrape Song of Songs
song = scrape_book('song_of_songs', verbose=True)
```

```python
# Basic sanity checks
print(f"Chapters scraped: {len(song)}")
for ch, verses in song.items():
    print(f"  Chapter {ch}: {len(verses)} verses")
```

```python
# Spot check — print first 5 verses of chapter 1
print("Chapter 1, first 5 verses:\n")
for v in song[1][:5]:
    print(f"  [{v['verse_number']}] {v['text']}")
```

```python
# Spot check — print first 5 verses of chapter 4 (poetic middle)
print("Chapter 4, first 5 verses:\n")
for v in song[4][:5]:
    print(f"  [{v['verse_number']}] {v['text']}")
```

```python
# Spot check — print first 5 verses of chapter 4 (poetic middle)
print("Chapter 5, first 5 verses:\n")
for v in song[5][:5]:
    print(f"  [{v['verse_number']}] {v['text']}")
```

```python
# Check for any suspiciously short or empty verses that slipped through
print("Verses shorter than 30 chars (potential noise):\n")
for ch, verses in song.items():
    for v in verses:
        if len(v['text']) < 30:
            print(f"  Ch {ch} v{v['verse_number']}: '{v['text']}'")
```
