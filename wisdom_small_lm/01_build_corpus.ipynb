# Build Wisdom Corpus
Scrapes all 7 wisdom books from BibleGateway (NRSVCE) and saves to `wisdom_corpus.json`.

| Book | Chapters |
|---|---|
| Psalms | 150 |
| Job | 42 |
| Proverbs | 31 |
| Ecclesiastes | 12 |
| Song of Songs | 8 |
| Wisdom | 19 |
| Sirach | 51 |
| Lamentations | 5 |
| **Total** | **318** |

Estimated time: ~5-7 minutes with 0.5s delay between requests.

```python
import sys
sys.path.insert(0, '.')
from scraper import scrape_all_books, save_corpus
```

```python
corpus = scrape_all_books(verbose=True)
```

```python
save_corpus(corpus, 'wisdom_corpus.json')
```

```python
# Summary
total_verses = 0
print(f"{'Book':<20} {'Chapters':>8} {'Verses':>8}")
print("-" * 38)
for book, chapters in corpus.items():
    verse_count = sum(len(v) for v in chapters.values())
    total_verses += verse_count
    print(f"{book.replace('_', ' ').title():<20} {len(chapters):>8} {verse_count:>8}")
print("-" * 38)
print(f"{'TOTAL':<20} {sum(len(c) for c in corpus.values()):>8} {total_verses:>8}")
```

```python
# Spot check — one random verse from each book
import random
random.seed(42)

from scraper import load_corpus
corpus = load_corpus('wisdom_corpus.json')

for book, chapters in corpus.items():
    ch_key = random.choice(list(chapters.keys()))
    verses = chapters[ch_key]
    if verses:
        v = random.choice(verses)
        label = f"{book.replace('_', ' ').title()} {ch_key}:{v['verse_number']}"
        print(f"[{label}]\n  {v['text']}\n")
```
